```python
df_encode = pd.get_dummies(df,drop_first=True)
```

```python
df_encode
```

```python
from sklearn.preprocessing import StandardScaler
numerical_cols = ['credit_score','age','tenure','balance','products_number','credit_card','estimated_salary']
scaler = StandardScaler()
df_encode[numerical_cols] = scaler.fit_transform(df_encode[numerical_cols])
df_encode.head()
```

```python
# 1. Define the boolean columns in df_encode
bool_cols = ["country_Germany", "country_Spain", "gender_Male"]

# 2. Map True to 1 and False to 0 for each column
for col in bool_cols:
    df_encode[col] = df_encode[col].map({True: 1, False: 0})

# 3. View the updated data
df_encode.head()
```

```python
df_encode.columns
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

# 1. Explicitly define your categorical/binary features
cat_features = [
    "active_member",
    "credit_card",
    "country_Germany",
    "country_Spain",
    "gender_Male",
]
alpha = 0.05
chi2_results = {}

# 2. Run Chi-Square test against 'churn'
for col in cat_features:
    # Build cross-tabulation table directly using the raw binary churn target
    contingency = pd.crosstab(df_encode[col], df_encode["churn"])

    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = (
        "Reject Null (Keep Feature)"
        if p_val < alpha
        else "Accept Null (Drop Feature)"
    )

    chi2_results[col] = {
        "chi2_statistic": chi2_stat,
        "p_value": p_val,
        "Decision": decision,
    }

# 3. Format and display results
chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by="p_value")
print("--- Chi-Square Test Results ---")
print(chi2_df)
```

```python
import pandas as pd
from scipy.stats import chi2_contingency, pearsonr

# Define the significance threshold
alpha = 0.05

# ==========================================
# PART 1: Chi-Square Test (Categorical/Binary)
# ==========================================
cat_features = [
    "credit_card",
    "active_member",
    "country_Germany",
    "country_Spain",
    "gender_Male",
]
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(df_encode[col], df_encode["churn"])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = (
        "Reject Null (Keep Feature)"
        if p_val < alpha
        else "Accept Null (Drop Feature)"
    )

    chi2_results[col] = {
        "Chi2 Statistic": chi2_stat,
        "P-Value": p_val,
        "Decision": decision,
    }

chi2_df = pd.DataFrame(chi2_results).T.sort_values(by="P-Value")

# ==========================================
# PART 2: Pearson Correlation (Continuous)
# ==========================================
numeric_features = [
    "credit_score",
    "age",
    "tenure",
    "balance",
    "products_number",
    "estimated_salary",
]
pearson_results = []

for feature in numeric_features:
    corr_coef, p_val = pearsonr(df_encode[feature], df_encode["churn"])
    significance = "Significant" if p_val < alpha else "Not Significant"

    pearson_results.append(
        {
            "Feature": feature,
            "Pearson Correlation": corr_coef,
            "P-Value": p_val,
            "Significance": significance,
        }
    )

correlation_df = pd.DataFrame(pearson_results).sort_values(
    by="Pearson Correlation", ascending=False
)

# ==========================================
# DISPLAY RESULTS
# ==========================================
print("=== CHI-SQUARE TEST RESULTS (Categorical Features) ===")
print(chi2_df.to_string())
print("\n" + "=" * 55 + "\n")
print("=== PEARSON CORRELATION RESULTS (Continuous Features) ===")
print(correlation_df.to_string(index=False))
```

```python
X = df_encode.drop('churn', axis=1)
y = df_encode['churn']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, stratify=y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
models = {
    "Logistic Regression": LogisticRegression(),
    "KNN": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(),
    "SVM (RBF Kernel)": SVC(probability=True)
}
```

```python
results = []
```

```python
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    results.append({
        'Model': name,
        'Accuracy': round(acc, 4),
        'F1 Score': round(f1, 4)
    })
```
